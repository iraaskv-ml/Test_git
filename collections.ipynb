### 1. Строка и словарь
Написать программу, которая получает на вход строку и возвращает словарь, где:
ключи — символы из этой строки;
значения — количество раз, сколько каждый символ встречается.

```python
s = input("Введите строку: ")
collection = {}

for char in s:
    if char in collection:
        collection[char] += 1
    else:
        collection[char] = 1

print(collection)
```

### 2. Обратный порядок строки
Дана строка текста (или введённая через консоль). Программа должна вернуть новую строку, в которой порядок слов будет обратным.

```python
s = input("Введите строку: ")

words = s.split()
words.reverse()

new_str = " ".join(words)

print(new_str)
```

### 3. Удаление дубликатов
Написать программу, которая удаляет из списка все дубликаты, сохранив исходный порядок элементов.

```python
num_set = [1, 5, 2, 5, 3, 1, 4, 2, 6]

new_set = []

for num in num_set:
    if num not in new_set:
        new_set.append(num)

print("Исходный список:", num_set)
print("Список без дубликатов:", new_set)
```

### 4. Список с уникальными элементами
Даны три (или больше) списка с объектами. Программа должна создать новый список, содержащий все уникальные элементы — каждый объект встречается только один раз.



```python
list1 = ['a', 'b', 'c', 'd', 'a']
list2 = [1, 2, 3, 4, 5, 1]
list3 = ['one', 'two', 'three', 'four', 'five', 'one' ]

big_list = []

for elements in list1 + list2 + list3:
    if elements not in big_list:
        big_list.append(elements)

print("Список с уникальными элементами:", big_list)
```

### 5. Разделение по словарю
Дана строка текста (или введённая через консоль). Программа должна вернуть словарь с четырьмя ключами:
"гласные",
"согласные",
"цифры",
"пунктуация".
Значения — количество символов каждого типа в строке.

```python
import string

s = input("Введите строку: ")

dist_dict = {
    "Char_gl": 0,
    "Char_sogl": 0,
    "Digit": 0,
    "Punct": 0
}

Char_gl = "аеёиоуыэюяaeiou"
Char_sogl = "бвгджзйклмнпрстфхцчшщbcdfghjklmnpqrstvwxyz"

for char in s:
    if char in Char_gl:
        dist_dict["Char_gl"] += 1
    elif char in Char_sogl:
        dist_dict["Char_sogl"] += 1
    elif char.isdigit():
        dist_dict["Digit"] += 1
    elif char in string.punctuation:
        dist_dict["Punct"] += 1

print(dist_dict)
```
